# B4: Design Rationale

I chose this topic because I have gone to many schools (preschool, elementary, middle, high, undergrad, and grad school) and the resources available to disabled students were different at each one. I have academic accommodations and it was always a challenge to move those accommodations from one school to another, and to figure out if the resources I use were even available at whatever school I was going to next. I want to create this dashboard so that anyone can view this data about the schools in their area and choose which ones to go to based on what accommodations are available. I also have never seen a dashboard like this before, but I read a GAO article calling for all this data to be updated frequently (https://www.gao.gov/products/gao-24-106264). I figure if I create this dashboard it will put more pressure on the Department of Education to enforce IDEA and mandate reporting from schools each year.

All my sources matter very much, and I would pull them from the Internet Archive if any of these federal agencies took them down. There may also be other agencies collecting this data, but if so, I did not find it. I will do a deeper dive in the coming weeks for alternative resources.

I thought about including school lunch options at each institution but since they depend on each student's IEP, I figured I would just share contact information from every school so users could speak with the schools directly about IEPs.
